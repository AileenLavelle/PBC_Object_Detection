## 2. Helper Functions

```python
def clear_memory():
    """Clear GPU and CPU memory between experiments."""
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
        torch.cuda.synchronize()


def save_progress(results, filepath):
    """Save experiment progress to JSON file."""
    clean_results = []
    for r in results:
        clean_r = {}
        for k, v in r.items():
            if isinstance(v, (int, float, str, bool, type(None))):
                clean_r[k] = v
            else:
                clean_r[k] = str(v)
        clean_results.append(clean_r)
    
    with open(filepath, 'w') as f:
        json.dump(clean_results, f, indent=2)


def load_progress(filepath):
    """Load previous experiment progress."""
    if Path(filepath).exists():
        with open(filepath, 'r') as f:
            return json.load(f)
    return []


def get_gpu_memory_usage():
    """Get current GPU memory usage."""
    if torch.cuda.is_available():
        used = torch.cuda.memory_allocated() / 1e9
        total = torch.cuda.get_device_properties(0).total_memory / 1e9
        return f"{used:.1f}/{total:.1f} GB"
    return "N/A"
```

```python
def evaluate_with_sahi(model_path, val_images_dir, slice_size, overlap_ratio=0.2, conf_thresh=0.25):
    """Evaluate a trained model using SAHI sliced inference."""
    if not SAHI_AVAILABLE:
        return None
    
    from sahi import AutoDetectionModel
    from sahi.predict import get_sliced_prediction
    
    try:
        detection_model = AutoDetectionModel.from_pretrained(
            model_type='ultralytics',
            model_path=str(model_path),
            confidence_threshold=conf_thresh,
            device='cuda:0' if torch.cuda.is_available() else 'cpu'
        )
        
        image_extensions = ['.jpg', '.jpeg', '.png', '.bmp', '.tiff']
        image_files = [f for f in Path(val_images_dir).iterdir() 
                       if f.suffix.lower() in image_extensions]
        
        total_detections = 0
        start_time = time.time()
        
        for img_path in image_files:
            result = get_sliced_prediction(
                str(img_path),
                detection_model,
                slice_height=slice_size,
                slice_width=slice_size,
                overlap_height_ratio=overlap_ratio,
                overlap_width_ratio=overlap_ratio,
            )
            total_detections += len(result.object_prediction_list)
        
        elapsed_time = time.time() - start_time
        
        del detection_model
        clear_memory()
        
        return {
            'total_detections': total_detections,
            'num_images': len(image_files),
            'time_per_image': elapsed_time / len(image_files) if image_files else 0
        }
    except Exception as e:
        print(f"   SAHI evaluation failed: {str(e)}")
        return None


def evaluate_standard(model_path, val_images_dir, imgsz, conf_thresh=0.25):
    """Evaluate a trained model using standard inference."""
    try:
        model = YOLO(str(model_path))
        
        image_extensions = ['.jpg', '.jpeg', '.png', '.bmp', '.tiff']
        image_files = [f for f in Path(val_images_dir).iterdir() 
                       if f.suffix.lower() in image_extensions]
        
        total_detections = 0
        start_time = time.time()
        
        for img_path in image_files:
            results = model.predict(
                str(img_path),
                imgsz=imgsz,
                conf=conf_thresh,
                verbose=False
            )
            total_detections += len(results[0].boxes)
        
        elapsed_time = time.time() - start_time
        
        del model
        clear_memory()
        
        return {
            'total_detections': total_detections,
            'num_images': len(image_files),
            'time_per_image': elapsed_time / len(image_files) if image_files else 0
        }
    except Exception as e:
        print(f"Standard evaluation failed: {str(e)}")
        return None
```

## 3. Define Large Image Experiments

```python
# Base configuration for large image experiments
BASE_CONFIG = {
    'data': data_yaml_path,
    'epochs': 100,
    'imgsz': 2048,
    'batch': 4,
    'rect': True,
    'device': 0,
    
    # Augmentations
    'mosaic': 0.5,
    'scale': 0.2,
    'fliplr': 0.5,
    'flipud': 0.0,
    
    # Training settings
    'patience': 30,
    'save': True,
    'plots': True,
    'project': 'jupiter_inlet_large_image',
    
    # Memory optimization
    'workers': 4,
}

# SAHI settings
SAHI_OVERLAP_RATIO = 0.2
SAHI_CONFIDENCE_THRESHOLD = 0.25
```

```python
# Define large image experiments
experiments = [
    # ==========================================================================
    # BASELINE AT 2048
    # ==========================================================================
    {
        'name': 'large_baseline_2048_100ep',
        'model': 'yolo26s.pt',
        'changes': {},
        'experiment_type': 'large_image'
    },
    
    # ==========================================================================
    # LARGER IMAGE SIZE
    # ==========================================================================
    {
        'name': 'large_imgsz_2560',
        'model': 'yolo26s.pt',
        'changes': {'imgsz': 2560, 'batch': 2},
        'experiment_type': 'large_image'
    },
    
    # ==========================================================================
    # MORE EPOCHS AT 2048
    # ==========================================================================
    {
        'name': 'large_2048_150ep',
        'model': 'yolo26s.pt',
        'changes': {'epochs': 150},
        'experiment_type': 'large_image'
    },
    {
        'name': 'large_2048_200ep',
        'model': 'yolo26s.pt',
        'changes': {'epochs': 200},
        'experiment_type': 'large_image'
    },
    
    # ==========================================================================
    # LARGER MODEL AT 2048
    # ==========================================================================
    {
        'name': 'large_medium_model_2048',
        'model': 'yolo26m.pt',
        'changes': {'batch': 2},
        'experiment_type': 'large_image'
    },
]

print(f"Total large image experiments: {len(experiments)}")
print("\nExperiment list:")
for i, exp in enumerate(experiments, 1):
    changes = exp['changes'] if exp['changes'] else 'baseline (2048, 100ep, yolo26s)'
    print(f"{i}. {exp['name']}: {exp['model']} | {changes}")
```

## 5. Check for Previous Progress

```python
# Check for previous progress on large image experiments
previous_large_results = load_progress(progress_file)

if previous_large_results:
    completed_experiments = [r['experiment'] for r in previous_large_results if r.get('status') == 'success']
    print(f"Found {len(completed_experiments)} completed large image experiments")
    for exp_name in completed_experiments:
        print(f" {exp_name}")
    
    RESUME_FROM_PREVIOUS = True
    
    if RESUME_FROM_PREVIOUS:
        results_summary = previous_large_results
        print(f"Resuming from previous run...")
    else:
        results_summary = []
        completed_experiments = []
else:
    results_summary = []
    completed_experiments = []
    print("No previous large image progress found. Starting fresh.")
```

```python
# Filter experiments
RUN_SAHI_EVALUATION = True and SAHI_AVAILABLE

experiments_to_run = [exp for exp in experiments 
                      if exp['name'] not in completed_experiments]

print(f"Will run {len(experiments_to_run)} experiment(s):")
for exp in experiments_to_run:
    print(f"  - {exp['name']}")

if not experiments_to_run:
    print("All large image experiments already completed!")
```

## 6. Run Large Image Experiments

```python
total_start_time = time.time()

print(f"Starting large image experiments at {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
print(f"GPU Memory at start: {get_gpu_memory_usage()}")
print("=" * 70)

for i, exp in enumerate(experiments_to_run, 1):
    print(f"\n{'='*70}")
    print(f"EXPERIMENT {i}/{len(experiments_to_run)}: {exp['name']}")
    print(f"Model: {exp['model']}")
    print(f"Changes: {exp['changes'] if exp['changes'] else 'baseline'}")
    print(f"GPU Memory: {get_gpu_memory_usage()}")
    print(f"{'='*70}\n")
    
    clear_memory()
    
    # Build config
    config = BASE_CONFIG.copy()
    config.update(exp['changes'])
    config['name'] = exp['name']
    
    imgsz = config['imgsz']
    exp_start_time = time.time()
    model = None
    
    try:
        # TRAINING
        print(f"Training with imgsz={imgsz}, batch={config['batch']}, epochs={config['epochs']}...")
        model = YOLO(exp['model'])
        results = model.train(**config)
        
        train_time = time.time() - exp_start_time
        
        metrics = results.results_dict
        best_model_path = Path(results.save_dir) / 'weights' / 'best.pt'
        
        result_entry = {
            'experiment': exp['name'],
            'model': exp['model'],
            'imgsz': imgsz,
            'epochs': config['epochs'],
            'batch': config['batch'],
            'mAP50': metrics.get('metrics/mAP50(B)', None),
            'mAP50-95': metrics.get('metrics/mAP50-95(B)', None),
            'precision': metrics.get('metrics/precision(B)', None),
            'recall': metrics.get('metrics/recall(B)', None),
            'training_time_min': round(train_time / 60, 1),
            'best_model_path': str(best_model_path),
            'experiment_type': 'large_image',
            'status': 'success'
        }
        
        print(f"Training completed in {train_time/60:.1f} min")
        print(f"   mAP50: {result_entry['mAP50']:.4f}")
        print(f"   Recall: {result_entry['recall']:.4f}")
        print(f"   Precision: {result_entry['precision']:.4f}")
        
        del model
        del results
        model = None
        clear_memory()
        
        # SAHI EVALUATION
        if RUN_SAHI_EVALUATION:
            print(f"\n   Running SAHI evaluation (slice_size={imgsz})...")
            
            std_results = evaluate_standard(
                best_model_path, val_images_dir, imgsz=imgsz,
                conf_thresh=SAHI_CONFIDENCE_THRESHOLD
            )
            
            sahi_results = evaluate_with_sahi(
                best_model_path, val_images_dir, slice_size=imgsz,
                overlap_ratio=SAHI_OVERLAP_RATIO, conf_thresh=SAHI_CONFIDENCE_THRESHOLD
            )
            
            if std_results:
                result_entry['std_detections'] = std_results['total_detections']
            
            if sahi_results:
                result_entry['sahi_detections'] = sahi_results['total_detections']
                result_entry['sahi_slice_size'] = imgsz
                
                if std_results and std_results['total_detections'] > 0:
                    detection_increase = sahi_results['total_detections'] - std_results['total_detections']
                    pct_increase = ((sahi_results['total_detections'] / std_results['total_detections']) - 1) * 100
                    result_entry['detection_increase'] = detection_increase
                    result_entry['detection_increase_pct'] = round(pct_increase, 1)
                    
                    print(f"   Standard: {std_results['total_detections']} | SAHI: {sahi_results['total_detections']} ({pct_increase:+.1f}%)")
        
    except Exception as e:
        exp_time = time.time() - exp_start_time
        print(f"FAILED: {exp['name']} after {exp_time/60:.1f} min")
        print(f"   Error: {str(e)}")
        
        result_entry = {
            'experiment': exp['name'],
            'model': exp['model'],
            'imgsz': imgsz,
            'epochs': config['epochs'],
            'batch': config['batch'],
            'mAP50': None,
            'recall': None,
            'precision': None,
            'training_time_min': round(exp_time / 60, 1),
            'experiment_type': 'large_image',
            'status': f'failed: {str(e)[:100]}'
        }
    
    finally:
        if model is not None:
            del model
        clear_memory()
    
    results_summary.append(result_entry)
    save_progress(results_summary, progress_file)
    print(f"\n   Progress saved. Completed {len(results_summary)}/{len(experiments)} experiments.")

total_time = time.time() - total_start_time
print(f"\n{'='*70}")
print(f"ALL LARGE IMAGE EXPERIMENTS COMPLETED")
print(f"Total time: {total_time/60:.1f} minutes ({total_time/3600:.1f} hours)")
print(f"{'='*70}")
```

## 7. SAHI Slice Size Comparison

Testing different SAHI slice sizes on the best model.

```python
# Find best model from large image experiments
successful_large = [r for r in results_summary if r.get('status') == 'success' and r.get('mAP50')]

if successful_large:
    best_large = max(successful_large, key=lambda x: x['mAP50'])
    print(f"Best large image model: {best_large['experiment']}")
    print(f"  mAP50: {best_large['mAP50']:.4f}")
    print(f"  Path: {best_large.get('best_model_path', 'N/A')}")
else:
    best_large = None
    print("No successful large image experiments yet.")
```

```python
# Test different SAHI slice sizes on best model
RUN_SAHI_SLICE_TEST = True and SAHI_AVAILABLE and best_large is not None

sahi_slice_results = []

if RUN_SAHI_SLICE_TEST:
    best_model_path = best_large.get('best_model_path')
    
    if best_model_path and Path(best_model_path).exists():
        slice_sizes = [512, 640, 768, 1024, 1280, 1536]
        
        print(f"\nTesting SAHI slice sizes on: {best_large['experiment']}")
        print("-" * 50)
        
        for slice_size in slice_sizes:
            print(f"Testing slice_size={slice_size}...", end=" ")
            
            result = evaluate_with_sahi(
                best_model_path, val_images_dir,
                slice_size=slice_size,
                overlap_ratio=SAHI_OVERLAP_RATIO,
                conf_thresh=SAHI_CONFIDENCE_THRESHOLD
            )
            
            if result:
                sahi_slice_results.append({
                    'slice_size': slice_size,
                    'detections': result['total_detections'],
                    'time_per_image': round(result['time_per_image'], 3)
                })
                print(f"Detections: {result['total_detections']}, Time: {result['time_per_image']:.3f}s/img")
            else:
                print("Failed")
        
        print("\nSAHI Slice Size Results:")
        print(f"{'Slice Size':<12} {'Detections':<12} {'Time/Img':<12}")
        print("-" * 36)
        for r in sahi_slice_results:
            print(f"{r['slice_size']:<12} {r['detections']:<12} {r['time_per_image']:<12}")
    else:
        print(f"Model path not found: {best_model_path}")
else:
    print("Skipping SAHI slice test.")
```

```python
# Plot 2: Image Size vs Performance Trend
if MATPLOTLIB_AVAILABLE and successful_all:
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Get experiments with valid imgsz and mAP50
    valid_exps = [r for r in successful_all if r.get('imgsz') and r.get('mAP50')]
    
    if valid_exps:
        # Sort by imgsz
        valid_exps_sorted = sorted(valid_exps, key=lambda x: x['imgsz'])
        
        imgsz_vals = [r['imgsz'] for r in valid_exps_sorted]
        mAP_vals = [r['mAP50'] for r in valid_exps_sorted]
        recall_vals = [r.get('recall', 0) or 0 for r in valid_exps_sorted]
        
        # Color by experiment type
        colors = ['coral' if r.get('experiment_type') == 'large_image' else 'steelblue' 
                  for r in valid_exps_sorted]
        
        # Left: mAP50 vs Image Size
        axes[0].scatter(imgsz_vals, mAP_vals, c=colors, s=100, alpha=0.7)
        axes[0].plot(imgsz_vals, mAP_vals, 'k--', alpha=0.3)
        axes[0].set_xlabel('Image Size', fontsize=12)
        axes[0].set_ylabel('mAP50', fontsize=12)
        axes[0].set_title('mAP50 vs Image Size', fontsize=14)
        axes[0].grid(True, alpha=0.3)
        
        # Add annotations for key points
        for r in valid_exps_sorted:
            if r['imgsz'] in [512, 1024, 2048, 2560]:
                axes[0].annotate(f"{r['mAP50']:.3f}", 
                                (r['imgsz'], r['mAP50']),
                                textcoords="offset points",
                                xytext=(0, 10), ha='center', fontsize=9)
        
        # Right: Recall vs Image Size
        axes[1].scatter(imgsz_vals, recall_vals, c=colors, s=100, alpha=0.7)
        axes[1].plot(imgsz_vals, recall_vals, 'k--', alpha=0.3)
        axes[1].set_xlabel('Image Size', fontsize=12)
        axes[1].set_ylabel('Recall', fontsize=12)
        axes[1].set_title('Recall vs Image Size', fontsize=14)
        axes[1].grid(True, alpha=0.3)
        
        # Legend
        from matplotlib.patches import Patch
        legend_elements = [
            Patch(facecolor='steelblue', label='Small Image (≤1024)'),
            Patch(facecolor='coral', label='Large Image (≥2048)')
        ]
        axes[0].legend(handles=legend_elements, loc='lower right')
    
    plt.tight_layout()
    
    fig_path = output_dir / 'imgsz_vs_performance.png'
    plt.savefig(fig_path, dpi=150, bbox_inches='tight')
    print(f"Saved: {fig_path}")
    
    plt.show()
```

```python
# Plot 3: Epochs vs Performance (for 2048 experiments)
if MATPLOTLIB_AVAILABLE:
    # Filter to 2048 experiments with different epochs
    epoch_exps = [r for r in successful_all 
                  if r.get('imgsz') == 2048 and r.get('epochs') and r.get('mAP50')]
    
    if len(epoch_exps) > 1:
        fig, ax = plt.subplots(figsize=(10, 5))
        
        epoch_exps_sorted = sorted(epoch_exps, key=lambda x: x['epochs'])
        
        epochs = [r['epochs'] for r in epoch_exps_sorted]
        mAPs = [r['mAP50'] for r in epoch_exps_sorted]
        recalls = [r.get('recall', 0) or 0 for r in epoch_exps_sorted]
        
        ax.plot(epochs, mAPs, 'o-', color='steelblue', markersize=12, linewidth=2, label='mAP50')
        ax.plot(epochs, recalls, 's-', color='forestgreen', markersize=12, linewidth=2, label='Recall')
        
        ax.set_xlabel('Epochs', fontsize=12)
        ax.set_ylabel('Score', fontsize=12)
        ax.set_title('Performance vs Epochs (imgsz=2048)', fontsize=14)
        ax.legend(fontsize=11)
        ax.grid(True, alpha=0.3)
        
        # Add value labels
        for e, m, r in zip(epochs, mAPs, recalls):
            ax.annotate(f'{m:.3f}', (e, m), textcoords="offset points", xytext=(0, 10), ha='center')
        
        plt.tight_layout()
        
        fig_path = output_dir / 'epochs_vs_performance_2048.png'
        plt.savefig(fig_path, dpi=150, bbox_inches='tight')
        print(f"Saved: {fig_path}")
        
        plt.show()
    else:
        print("Not enough epoch experiments at 2048 to plot.")
```

```python
# Plot 4: SAHI Slice Size Comparison
if MATPLOTLIB_AVAILABLE and sahi_slice_results:
    fig, ax1 = plt.subplots(figsize=(10, 5))
    
    slice_sizes = [r['slice_size'] for r in sahi_slice_results]
    detections = [r['detections'] for r in sahi_slice_results]
    times = [r['time_per_image'] for r in sahi_slice_results]
    
    # Detections
    color1 = 'steelblue'
    ax1.plot(slice_sizes, detections, 'o-', color=color1, markersize=12, linewidth=2)
    ax1.set_xlabel('SAHI Slice Size', fontsize=12)
    ax1.set_ylabel('Total Detections', color=color1, fontsize=12)
    ax1.tick_params(axis='y', labelcolor=color1)
    
    # Time (secondary axis)
    ax2 = ax1.twinx()
    color2 = 'coral'
    ax2.plot(slice_sizes, times, 's--', color=color2, markersize=10, linewidth=2)
    ax2.set_ylabel('Time per Image (s)', color=color2, fontsize=12)
    ax2.tick_params(axis='y', labelcolor=color2)
    
    plt.title('SAHI Performance vs Slice Size', fontsize=14)
    
    # Find optimal (most detections)
    best_idx = detections.index(max(detections))
    ax1.annotate(f'Best: {slice_sizes[best_idx]}',
                xy=(slice_sizes[best_idx], detections[best_idx]),
                xytext=(10, 20), textcoords='offset points',
                fontsize=11, color='green',
                arrowprops=dict(arrowstyle='->', color='green'))
    
    plt.tight_layout()
    
    fig_path = output_dir / 'sahi_slice_comparison.png'
    plt.savefig(fig_path, dpi=150, bbox_inches='tight')
    print(f"Saved: {fig_path}")
    
    plt.show()
```

```python
# Plot 5: Combined Overview - All Experiments Ranked
if MATPLOTLIB_AVAILABLE and sorted_results:
    fig, ax = plt.subplots(figsize=(12, max(6, len(sorted_results) * 0.4)))
    
    names = [r['experiment'][:25] for r in sorted_results]
    mAPs = [r['mAP50'] for r in sorted_results]
    types = [r.get('experiment_type', 'unknown') for r in sorted_results]
    
    # Color by type
    colors = []
    for i, t in enumerate(types):
        if i == 0:
            colors.append('gold')  # Best overall
        elif t == 'large_image':
            colors.append('coral')
        else:
            colors.append('steelblue')
    
    bars = ax.barh(range(len(names)), mAPs, color=colors)
    ax.set_yticks(range(len(names)))
    ax.set_yticklabels(names)
    ax.set_xlabel('mAP50', fontsize=12)
    ax.set_title('All Experiments Ranked by mAP50', fontsize=14)
    ax.set_xlim(0, 1)
    
    # Add value labels
    for i, v in enumerate(mAPs):
        ax.text(v + 0.01, i, f'{v:.3f}', va='center', fontsize=9)
    
    # Legend
    from matplotlib.patches import Patch
    legend_elements = [
        Patch(facecolor='gold', label='Best Overall'),
        Patch(facecolor='coral', label='Large Image (≥2048)'),
        Patch(facecolor='steelblue', label='Small Image (≤1024)')
    ]
    ax.legend(handles=legend_elements, loc='lower right')
    
    plt.tight_layout()
    
    fig_path = output_dir / 'all_experiments_ranked.png'
    plt.savefig(fig_path, dpi=150, bbox_inches='tight')
    print(f"Saved: {fig_path}")
    
    plt.show()
```
